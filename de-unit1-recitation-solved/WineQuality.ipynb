```python
import pandas as pd
```

```python
df = pd.read_csv('WineQT.csv')
```

```python
df.head()
```

```python
# Treat zeros as missing for this exercise.
df["citric acid"] = df["citric acid"].replace(0.00, float("nan"))
```

```python
df.head()
```

```python
#1.	Fill all null values in citric acid column with mean value of the column
df["citric acid"] = df["citric acid"].fillna(df["citric acid"].mean())
```

```python
df.head()
```

```python
# 2. Using iloc - Print “alcohol” values for the first 10 rows
# alternative ways to get the first 10 "alcohol" values
df["alcohol"].iloc[:10]
#df.iloc[:10, df.columns.get_loc("alcohol")]
#df.alcohol.iloc[:10]

#df.alcohol[:10]
#df.loc[:9, "alcohol"]
#df.head(10)["alcohol"]
```

```python
# 3.	Using loc - Select rows where alcohol > 10 AND show only alcohol + quality columns
df.loc[df["alcohol"] > 10, ["alcohol", "quality"]]
# alternative ways
#df[df["alcohol"] > 10][["alcohol", "quality"]]
#df.query("alcohol > 10")[["alcohol", "quality"]]
#df.loc[df.alcohol > 10, ["alcohol", "quality"]]
```

```python
# 4.	Create a new dataframe such that you want to know the average alcohol content , minimum fixed acidity and maximum fixed acidity by the quality rating of wine (use group by)
new_df = df.groupby("quality").agg({
    "alcohol": "mean",
    "fixed acidity": ["min", "max"]
})
new_df
```

```python
# Sort the values by alcohol mean
new_df.sort_values(("alcohol", "mean"))
```

```python
#5.	Do the same using pivot table
pt = pd.pivot_table(
    df,
    index="quality",
    values=["alcohol", "fixed acidity"],
    aggfunc={"alcohol": "mean", "fixed acidity": ["min", "max"]}
)

# Sort by alcohol mean
pt.sort_values(("alcohol", "mean"))
```

```python
movies = pd.DataFrame({
    'movie_id': [1, 2, 3, 4],
    'title': ['Inception', 'The Matrix', 'Avengers', 'Interstellar'],
    'genre': ['Sci-Fi', 'Sci-Fi', 'Action', 'Sci-Fi']
})
```

```python
ratings = pd.DataFrame({
    'movie_id': [1, 2, 3, 5],
    'rating': [8.8, 8.7, 8.4, 7.5],
    'votes': [2000000, 1500000, 2200000, 500000]
})
```

```python
#6.	Merge this dataframe to the current dataframe on movie_id (inner, left and outer join)
merged_df1 = movies.merge(ratings, on='movie_id', how='inner')
merged_df1
```

```python
merged_df = movies.merge(ratings, on='movie_id', how='left')
print(merged_df)
```

```python
merged_df = movies.merge(ratings, on='movie_id', how='outer')
print(merged_df)
```
